## 1. What is EDA and Why It Matters

### Definition

**Exploratory Data Analysis (EDA)** is the process of investigating datasets to:
- Understand the data structure
- Discover patterns and relationships
- Identify anomalies and outliers
- Test assumptions
- Generate hypotheses

### Why EDA is Critical

```
"Garbage In, Garbage Out"
```

No matter how sophisticated your model is, if you don't understand your data:
- ❌ You'll make wrong assumptions
- ❌ You'll miss important patterns
- ❌ You'll build models that don't work
- ❌ You'll waste time debugging later

**Good EDA = Foundation for Good Models**

## 2. First Look at Data

### Loading and Initial Inspection

Let's create a realistic dataset to work with:

````python
# Create a realistic customer dataset
np.random.seed(42)
n = 1000

df = pd.DataFrame({
    'customer_id': range(1, n+1),
    'age': np.random.randint(18, 80, n),
    'gender': np.random.choice(['Male', 'Female', 'Other'], n, p=[0.48, 0.48, 0.04]),
    'income': np.random.lognormal(10.5, 0.5, n).astype(int),
    'purchase_amount': np.random.gamma(2, 50, n),
    'num_purchases': np.random.poisson(5, n),
    'membership_years': np.random.uniform(0, 10, n),
    'satisfaction_score': np.random.choice([1, 2, 3, 4, 5], n, p=[0.05, 0.10, 0.30, 0.35, 0.20]),
    'city': np.random.choice(['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix'], n),
    'product_category': np.random.choice(['Electronics', 'Clothing', 'Food', 'Books'], n)
})

# Introduce some missing values (realistic!)
missing_indices = np.random.choice(df.index, size=50, replace=False)
df.loc[missing_indices, 'income'] = np.nan

missing_indices2 = np.random.choice(df.index, size=30, replace=False)
df.loc[missing_indices2, 'satisfaction_score'] = np.nan

# Add some outliers
outlier_indices = np.random.choice(df.index, size=10, replace=False)
df.loc[outlier_indices, 'purchase_amount'] *= 10

print("✅ Dataset created!")
print(f"\nShape: {df.shape[0]} rows × {df.shape[1]} columns")
````

**Univariate** = Analyzing ONE variable at a time

### 4.1 Numerical Variables

````python
print("="*70)
print("UNIVARIATE ANALYSIS: NUMERICAL VARIABLES")
print("="*70)

# Pick a numerical column to analyze
col = 'age'

print(f"\n📊 Analyzing: {col}")
print(f"\nBasic Statistics:")
print(f"  Count:    {df[col].count()}")
print(f"  Missing:  {df[col].isnull().sum()} ({df[col].isnull().sum()/len(df)*100:.1f}%)")
print(f"  Mean:     {df[col].mean():.2f}")
print(f"  Median:   {df[col].median():.2f}")
print(f"  Std Dev:  {df[col].std():.2f}")
print(f"  Min:      {df[col].min():.2f}")
print(f"  Max:      {df[col].max():.2f}")
print(f"  Range:    {df[col].max() - df[col].min():.2f}")
print(f"  Skewness: {df[col].skew():.2f}")
print(f"  Kurtosis: {df[col].kurtosis():.2f}")

# Visualize
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# 1. Histogram
axes[0, 0].hist(df[col].dropna(), bins=30, edgecolor='black', alpha=0.7)
axes[0, 0].axvline(df[col].mean(), color='red', linestyle='--', linewidth=2, label=f'Mean: {df[col].mean():.1f}')
axes[0, 0].axvline(df[col].median(), color='green', linestyle='--', linewidth=2, label=f'Median: {df[col].median():.1f}')
axes[0, 0].set_xlabel(col)
axes[0, 0].set_ylabel('Frequency')
axes[0, 0].set_title(f'Histogram: {col}')
axes[0, 0].legend()
axes[0, 0].grid(axis='y', alpha=0.3)

# 2. Box Plot
axes[0, 1].boxplot(df[col].dropna(), vert=True)
axes[0, 1].set_ylabel(col)
axes[0, 1].set_title(f'Box Plot: {col}')
axes[0, 1].grid(axis='y', alpha=0.3)

# 3. KDE (Kernel Density Estimate)
df[col].dropna().plot(kind='kde', ax=axes[1, 0], linewidth=2)
axes[1, 0].set_xlabel(col)
axes[1, 0].set_ylabel('Density')
axes[1, 0].set_title(f'Density Plot: {col}')
axes[1, 0].grid(True, alpha=0.3)

# 4. Q-Q Plot (check normality)
stats.probplot(df[col].dropna(), dist="norm", plot=axes[1, 1])
axes[1, 1].set_title(f'Q-Q Plot: {col} (Normality Check)')
axes[1, 1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("\n💡 Interpretation:")
print("   • Histogram: Shows distribution shape")
print("   • Box Plot: Shows quartiles, median, outliers")
print("   • Density Plot: Smooth version of histogram")
print("   • Q-Q Plot: Points on line = normally distributed")
````

### Complete Univariate Analysis for All Columns

````python
def univariate_summary(df, column):
    """
    Comprehensive univariate analysis for any column.
    """
    print(f"\n{'='*70}")
    print(f"COLUMN: {column}")
    print(f"{'='*70}")
    
    # Basic info
    dtype = df[column].dtype
    n_total = len(df)
    n_missing = df[column].isnull().sum()
    n_unique = df[column].nunique()
    
    print(f"\nType: {dtype}")
    print(f"Total values: {n_total:,}")
    print(f"Missing: {n_missing:,} ({n_missing/n_total*100:.1f}%)")
    print(f"Unique values: {n_unique:,}")
    
    # Numerical
    if dtype in ['int64', 'float64']:
        print(f"\nStatistics:")
        print(f"  Mean:   {df[column].mean():.2f}")
        print(f"  Median: {df[column].median():.2f}")
        print(f"  Std:    {df[column].std():.2f}")
        print(f"  Min:    {df[column].min():.2f}")
        print(f"  Max:    {df[column].max():.2f}")
        
        # Check for outliers (IQR method)
        Q1 = df[column].quantile(0.25)
        Q3 = df[column].quantile(0.75)
        IQR = Q3 - Q1
        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR
        outliers = ((df[column] < lower) | (df[column] > upper)).sum()
        print(f"\n  Outliers: {outliers} ({outliers/n_total*100:.1f}%)")
    
    # Categorical
    elif dtype == 'object':
        print(f"\nTop 5 values:")
        top_5 = df[column].value_counts().head()
        for val, count in top_5.items():
            print(f"  {val}: {count} ({count/n_total*100:.1f}%)")

# Run for key columns
key_columns = ['age', 'income', 'purchase_amount', 'gender', 'city']

for col in key_columns:
    if col in df.columns:
        univariate_summary(df, col)
````

## 5. Bivariate Analysis

**Bivariate** = Analyzing relationships between TWO variables

### 5.1 Numerical vs Numerical

````python
print("="*70)
print("BIVARIATE: NUMERICAL vs NUMERICAL")
print("="*70)

var1, var2 = 'income', 'purchase_amount'

print(f"\nAnalyzing relationship: {var1} vs {var2}")

# Calculate correlation
correlation = df[[var1, var2]].corr().iloc[0, 1]

print(f"\nCorrelation: {correlation:.3f}")
if abs(correlation) > 0.7:
    strength = "Strong"
elif abs(correlation) > 0.3:
    strength = "Moderate"
else:
    strength = "Weak"
    
direction = "Positive" if correlation > 0 else "Negative"
print(f"Interpretation: {strength} {direction} relationship")

# Visualize
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# 1. Scatter plot
axes[0].scatter(df[var1], df[var2], alpha=0.5, s=30)
axes[0].set_xlabel(var1)
axes[0].set_ylabel(var2)
axes[0].set_title(f'Scatter Plot: {var1} vs {var2}\nCorrelation: {correlation:.3f}')
axes[0].grid(True, alpha=0.3)

# Add trend line
z = np.polyfit(df[var1].dropna(), df[var2].dropna(), 1)
p = np.poly1d(z)
x_line = np.linspace(df[var1].min(), df[var1].max(), 100)
axes[0].plot(x_line, p(x_line), "r--", linewidth=2, label='Trend line')
axes[0].legend()

# 2. Hexbin plot (better for large datasets)
axes[1].hexbin(df[var1].dropna(), df[var2].dropna(), gridsize=20, cmap='YlOrRd')
axes[1].set_xlabel(var1)
axes[1].set_ylabel(var2)
axes[1].set_title(f'Hexbin Plot: {var1} vs {var2}\n(Darker = More points)')
plt.colorbar(axes[1].collections[0], ax=axes[1], label='Count')

plt.tight_layout()
plt.show()

print("\n💡 What to look for:")
print("   • Linear relationship? (points form a line)")
print("   • Non-linear patterns? (curve, clusters)")
print("   • Outliers? (points far from others)")
print("   • Constant variance? (spread consistent)")
````

### 5.2 Categorical vs Numerical

````python
print("="*70)
print("BIVARIATE: CATEGORICAL vs NUMERICAL")
print("="*70)

cat_var, num_var = 'gender', 'purchase_amount'

print(f"\nAnalyzing: {num_var} by {cat_var}")

# Group statistics
group_stats = df.groupby(cat_var)[num_var].agg(['count', 'mean', 'median', 'std'])
print(f"\nStatistics by {cat_var}:")
print(group_stats)

# Visualize
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# 1. Box plot
df.boxplot(column=num_var, by=cat_var, ax=axes[0, 0])
axes[0, 0].set_xlabel(cat_var)
axes[0, 0].set_ylabel(num_var)
axes[0, 0].set_title(f'Box Plot: {num_var} by {cat_var}')
plt.sca(axes[0, 0])
plt.xticks(rotation=45, ha='right')

# 2. Violin plot
for i, category in enumerate(df[cat_var].unique()):
    data = df[df[cat_var] == category][num_var].dropna()
    parts = axes[0, 1].violinplot([data], positions=[i], showmeans=True, showmedians=True)
axes[0, 1].set_xticks(range(len(df[cat_var].unique())))
axes[0, 1].set_xticklabels(df[cat_var].unique(), rotation=45, ha='right')
axes[0, 1].set_xlabel(cat_var)
axes[0, 1].set_ylabel(num_var)
axes[0, 1].set_title(f'Violin Plot: {num_var} by {cat_var}')
axes[0, 1].grid(axis='y', alpha=0.3)

# 3. Bar plot (means)
group_stats['mean'].plot(kind='bar', ax=axes[1, 0], color='steelblue', edgecolor='black', alpha=0.7)
axes[1, 0].set_xlabel(cat_var)
axes[1, 0].set_ylabel(f'Mean {num_var}')
axes[1, 0].set_title(f'Mean {num_var} by {cat_var}')
axes[1, 0].grid(axis='y', alpha=0.3)
plt.setp(axes[1, 0].xaxis.get_majorticklabels(), rotation=45, ha='right')

# 4. Overlapping histograms
for category in df[cat_var].unique():
    data = df[df[cat_var] == category][num_var].dropna()
    axes[1, 1].hist(data, bins=20, alpha=0.5, label=category, edgecolor='black')
axes[1, 1].set_xlabel(num_var)
axes[1, 1].set_ylabel('Frequency')
axes[1, 1].set_title(f'Distribution of {num_var} by {cat_var}')
axes[1, 1].legend()
axes[1, 1].grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()

print("\n💡 What to look for:")
print("   • Do groups have different distributions?")
print("   • Is one group significantly higher/lower?")
print("   • Different spread (variance) in groups?")
print("   • Outliers in specific groups?")
````

### 5.3 Categorical vs Categorical

````python
print("="*70)
print("BIVARIATE: CATEGORICAL vs CATEGORICAL")
print("="*70)

var1, var2 = 'gender', 'product_category'

print(f"\nAnalyzing: {var1} vs {var2}")

# Cross-tabulation
crosstab = pd.crosstab(df[var1], df[var2])
print(f"\nCross-tabulation (counts):")
print(crosstab)

# Percentages
crosstab_pct = pd.crosstab(df[var1], df[var2], normalize='index') * 100
print(f"\nPercentages (row-wise):")
print(crosstab_pct.round(1))

# Chi-square test
from scipy.stats import chi2_contingency
chi2, p_value, dof, expected = chi2_contingency(crosstab)
print(f"\nChi-square test:")
print(f"  Chi-square statistic: {chi2:.2f}")
print(f"  P-value: {p_value:.4f}")
if p_value < 0.05:
    print(f"  → Variables are DEPENDENT (p < 0.05)")
else:
    print(f"  → Variables are INDEPENDENT (p >= 0.05)")

# Visualize
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# 1. Stacked bar chart
crosstab.plot(kind='bar', stacked=True, ax=axes[0], colormap='viridis', edgecolor='black')
axes[0].set_xlabel(var1)
axes[0].set_ylabel('Count')
axes[0].set_title(f'Stacked Bar: {var1} vs {var2}')
axes[0].legend(title=var2)
axes[0].grid(axis='y', alpha=0.3)
plt.setp(axes[0].xaxis.get_majorticklabels(), rotation=45, ha='right')

# 2. Heatmap
sns.heatmap(crosstab, annot=True, fmt='d', cmap='YlOrRd', ax=axes[1], cbar_kws={'label': 'Count'})
axes[1].set_xlabel(var2)
axes[1].set_ylabel(var1)
axes[1].set_title(f'Heatmap: {var1} vs {var2}')

plt.tight_layout()
plt.show()

print("\n💡 What to look for:")
print("   • Are categories independent or related?")
print("   • Dominant combinations?")
print("   • Unexpected patterns?")
````

## 6. Multivariate Analysis

**Multivariate** = Analyzing relationships among MULTIPLE variables

### 6.1 Correlation Matrix

````python
print("="*70)
print("CORRELATION MATRIX")
print("="*70)

# Select numerical columns
numerical_df = df[numerical_cols]

# Calculate correlation matrix
corr_matrix = numerical_df.corr()

print("\nCorrelation Matrix:")
print(corr_matrix.round(3))

# Visualize
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', center=0, 
            square=True, linewidths=1, cbar_kws={"shrink": 0.8}, fmt='.2f')
plt.title('Correlation Matrix Heatmap', fontsize=14, pad=20)
plt.tight_layout()
plt.show()

# Find strong correlations
print("\n🔍 Strong correlations (|r| > 0.5):")
strong_corr = []
for i in range(len(corr_matrix.columns)):
    for j in range(i+1, len(corr_matrix.columns)):
        if abs(corr_matrix.iloc[i, j]) > 0.5:
            strong_corr.append((
                corr_matrix.columns[i], 
                corr_matrix.columns[j], 
                corr_matrix.iloc[i, j]
            ))

if strong_corr:
    for var1, var2, corr in strong_corr:
        print(f"   {var1} ↔ {var2}: {corr:.3f}")
else:
    print("   No strong correlations found")

print("\n💡 Interpretation:")
print("   • +1: Perfect positive correlation")
print("   •  0: No correlation")
print("   • -1: Perfect negative correlation")
print("   • |r| > 0.7: Strong")
print("   • |r| 0.3-0.7: Moderate")
print("   • |r| < 0.3: Weak")
````

````python
print("="*70)
print("MISSING VALUES ANALYSIS")
print("="*70)

# Count missing values
missing_count = df.isnull().sum()
missing_pct = (missing_count / len(df)) * 100

missing_df = pd.DataFrame({
    'Missing_Count': missing_count,
    'Missing_Percentage': missing_pct
}).sort_values('Missing_Count', ascending=False)

# Show only columns with missing values
missing_df = missing_df[missing_df['Missing_Count'] > 0]

print("\nMissing Values Summary:")
if len(missing_df) > 0:
    print(missing_df)
else:
    print("   ✅ No missing values!")

# Visualize
if len(missing_df) > 0:
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    
    # Bar plot
    missing_df['Missing_Percentage'].plot(kind='barh', ax=axes[0], color='coral', edgecolor='black')
    axes[0].set_xlabel('Missing Percentage (%)')
    axes[0].set_title('Missing Values by Column')
    axes[0].grid(axis='x', alpha=0.3)
    
    # Missing value matrix
    missing_matrix = df[missing_df.index].isnull().astype(int)
    sns.heatmap(missing_matrix.T, cmap='RdYlGn_r', cbar=False, ax=axes[1])
    axes[1].set_xlabel('Sample Index')
    axes[1].set_ylabel('Columns with Missing Values')
    axes[1].set_title('Missing Value Pattern\n(Red = Missing)')
    
    plt.tight_layout()
    plt.show()

print("\n💡 Decision Guide:")
print("   • < 5% missing: Usually safe to remove rows")
print("   • 5-20% missing: Impute (mean, median, mode)")
print("   • > 20% missing: Consider dropping column or advanced imputation")
print("   • Always check if missing is random or has a pattern!")
````

### 7.2 Duplicate Detection

````python
print("="*70)
print("DUPLICATE DETECTION")
print("="*70)

# Check for duplicates
n_duplicates = df.duplicated().sum()
pct_duplicates = (n_duplicates / len(df)) * 100

print(f"\nTotal rows: {len(df):,}")
print(f"Duplicate rows: {n_duplicates:,} ({pct_duplicates:.2f}%)")

if n_duplicates > 0:
    print(f"\n⚠️  Found {n_duplicates} duplicate rows!")
    print(f"\nExample duplicates:")
    display(df[df.duplicated(keep=False)].head(10))
else:
    print("\n✅ No duplicate rows found!")

# Check for duplicate IDs
if 'customer_id' in df.columns:
    n_dup_ids = df['customer_id'].duplicated().sum()
    print(f"\nDuplicate customer IDs: {n_dup_ids}")
    if n_dup_ids > 0:
        print("   ⚠️  Warning: IDs should be unique!")
````

### 7.3 Data Type Issues

````python
print("="*70)
print("DATA TYPE VALIDATION")
print("="*70)

print("\nCurrent data types:")
print(df.dtypes)

print("\n🔍 Checking for issues:")

issues_found = False

# Check numerical columns for non-numeric values
for col in numerical_cols:
    if df[col].dtype == 'object':
        print(f"   ⚠️  {col}: Should be numeric but is 'object' type")
        issues_found = True

# Check for negative values where they shouldn't be
non_negative_cols = ['age', 'income', 'purchase_amount', 'num_purchases']
for col in non_negative_cols:
    if col in df.columns and (df[col] < 0).any():
        n_negative = (df[col] < 0).sum()
        print(f"   ⚠️  {col}: Has {n_negative} negative values")
        issues_found = True

# Check for unrealistic values
if 'age' in df.columns:
    unrealistic_age = ((df['age'] < 0) | (df['age'] > 120)).sum()
    if unrealistic_age > 0:
        print(f"   ⚠️  age: {unrealistic_age} unrealistic values (< 0 or > 120)")
        issues_found = True

if not issues_found:
    print("   ✅ No obvious data type issues found!")

print("\n💡 Common fixes:")
print("   • Convert to numeric: pd.to_numeric(df['col'], errors='coerce')")
print("   • Convert to datetime: pd.to_datetime(df['col'])")
print("   • Convert to category: df['col'].astype('category')")
````

## 8. Outlier Detection and Treatment

````python
print("="*70)
print("OUTLIER DETECTION")
print("="*70)

def detect_outliers(df, column):
    """
    Detect outliers using IQR and Z-score methods.
    """
    data = df[column].dropna()
    
    # IQR method
    Q1 = data.quantile(0.25)
    Q3 = data.quantile(0.75)
    IQR = Q3 - Q1
    lower_iqr = Q1 - 1.5 * IQR
    upper_iqr = Q3 + 1.5 * IQR
    outliers_iqr = ((data < lower_iqr) | (data > upper_iqr))
    
    # Z-score method
    z_scores = np.abs(stats.zscore(data))
    outliers_z = z_scores > 3
    
    return {
        'iqr_count': outliers_iqr.sum(),
        'iqr_pct': outliers_iqr.sum() / len(data) * 100,
        'z_count': outliers_z.sum(),
        'z_pct': outliers_z.sum() / len(data) * 100,
        'lower_bound': lower_iqr,
        'upper_bound': upper_iqr
    }

# Analyze outliers for each numerical column
print("\nOutlier Analysis:")
print(f"\n{'Column':<20} {'IQR Method':<20} {'Z-Score Method'}")
print("-"*70)

for col in numerical_cols:
    result = detect_outliers(df, col)
    print(f"{col:<20} {result['iqr_count']:>5} ({result['iqr_pct']:>5.1f}%)     "
          f"{result['z_count']:>5} ({result['z_pct']:>5.1f}%)")

# Visualize outliers for one column
col_to_analyze = 'purchase_amount'

result = detect_outliers(df, col_to_analyze)

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# 1. Box plot
axes[0, 0].boxplot(df[col_to_analyze].dropna())
axes[0, 0].set_ylabel(col_to_analyze)
axes[0, 0].set_title(f'Box Plot: {col_to_analyze}\n(Points outside whiskers = outliers)')
axes[0, 0].grid(axis='y', alpha=0.3)

# 2. Histogram with outlier bounds
axes[0, 1].hist(df[col_to_analyze].dropna(), bins=30, edgecolor='black', alpha=0.7)
axes[0, 1].axvline(result['lower_bound'], color='red', linestyle='--', 
                   linewidth=2, label=f"Lower: {result['lower_bound']:.1f}")
axes[0, 1].axvline(result['upper_bound'], color='red', linestyle='--', 
                   linewidth=2, label=f"Upper: {result['upper_bound']:.1f}")
axes[0, 1].set_xlabel(col_to_analyze)
axes[0, 1].set_ylabel('Frequency')
axes[0, 1].set_title(f'Histogram with IQR Bounds')
axes[0, 1].legend()
axes[0, 1].grid(axis='y', alpha=0.3)

# 3. Scatter plot with index
data = df[col_to_analyze].dropna()
outliers = ((data < result['lower_bound']) | (data > result['upper_bound']))
colors = ['red' if x else 'blue' for x in outliers]
axes[1, 0].scatter(range(len(data)), data, c=colors, alpha=0.6, s=30)
axes[1, 0].axhline(result['lower_bound'], color='red', linestyle='--', alpha=0.5)
axes[1, 0].axhline(result['upper_bound'], color='red', linestyle='--', alpha=0.5)
axes[1, 0].set_xlabel('Index')
axes[1, 0].set_ylabel(col_to_analyze)
axes[1, 0].set_title(f'Scatter: Outliers in Red ({outliers.sum()} points)')
axes[1, 0].grid(True, alpha=0.3)

# 4. Summary statistics comparison
stats_with = df[col_to_analyze].describe()
stats_without = df[col_to_analyze][~outliers].describe()

comparison = pd.DataFrame({
    'With Outliers': stats_with,
    'Without Outliers': stats_without
})

axes[1, 1].axis('off')
table = axes[1, 1].table(cellText=comparison.round(2).values,
                          rowLabels=comparison.index,
                          colLabels=comparison.columns,
                          cellLoc='right',
                          loc='center')
table.auto_set_font_size(False)
table.set_fontsize(9)
table.scale(1, 2)
axes[1, 1].set_title('Statistics Comparison')

plt.tight_layout()
plt.show()

print("\n💡 Outlier Treatment Options:")
print("   1. Remove: df = df[~outliers]")
print("   2. Cap: df[col].clip(lower=lower_bound, upper=upper_bound)")
print("   3. Transform: np.log(df[col]) or np.sqrt(df[col])")
print("   4. Separate model: Train separate model for outliers")
print("   5. Keep: If they're valid data points!")
````

## 9. Complete EDA Workflow Template

````python
def complete_eda(df, target_column=None):
    """
    Complete EDA workflow for any dataset.
    
    Parameters:
    -----------
    df : DataFrame
        Dataset to analyze
    target_column : str, optional
        Target variable for supervised learning
    """
    print("\n" + "="*70)
    print("COMPLETE EDA REPORT")
    print("="*70)
    
    # 1. Basic Information
    print("\n1. DATASET OVERVIEW")
    print("-"*70)
    print(f"   Shape: {df.shape[0]:,} rows × {df.shape[1]} columns")
    print(f"   Memory: {df.memory_usage(deep=True).sum() / 1024**2:.2f} MB")
    
    # 2. Data Types
    print("\n2. DATA TYPES")
    print("-"*70)
    print(df.dtypes.value_counts())
    
    # 3. Missing Values
    print("\n3. MISSING VALUES")
    print("-"*70)
    missing = df.isnull().sum()
    missing_pct = (missing / len(df) * 100).round(2)
    missing_df = pd.DataFrame({
        'Missing': missing[missing > 0],
        'Percentage': missing_pct[missing > 0]
    }).sort_values('Missing', ascending=False)
    
    if len(missing_df) > 0:
        print(missing_df)
    else:
        print("   ✅ No missing values")
    
    # 4. Duplicates
    print("\n4. DUPLICATES")
    print("-"*70)
    n_dup = df.duplicated().sum()
    print(f"   Duplicate rows: {n_dup} ({n_dup/len(df)*100:.2f}%)")
    
    # 5. Numerical Summary
    print("\n5. NUMERICAL VARIABLES SUMMARY")
    print("-"*70)
    numerical = df.select_dtypes(include=['number']).columns
    if len(numerical) > 0:
        print(df[numerical].describe())
    else:
        print("   No numerical columns")
    
    # 6. Categorical Summary
    print("\n6. CATEGORICAL VARIABLES SUMMARY")
    print("-"*70)
    categorical = df.select_dtypes(include=['object']).columns
    if len(categorical) > 0:
        for col in categorical:
            n_unique = df[col].nunique()
            print(f"   {col}: {n_unique} unique values")
            if n_unique <= 10:
                print(df[col].value_counts().head())
                print()
    else:
        print("   No categorical columns")
    
    # 7. Outliers
    print("\n7. OUTLIER DETECTION (IQR Method)")
    print("-"*70)
    for col in numerical:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        outliers = ((df[col] < Q1 - 1.5*IQR) | (df[col] > Q3 + 1.5*IQR)).sum()
        if outliers > 0:
            print(f"   {col}: {outliers} outliers ({outliers/len(df)*100:.1f}%)")
    
    # 8. Target variable (if provided)
    if target_column and target_column in df.columns:
        print(f"\n8. TARGET VARIABLE: {target_column}")
        print("-"*70)
        if df[target_column].dtype in ['int64', 'float64']:
            print(df[target_column].describe())
        else:
            print(df[target_column].value_counts())
    
    print("\n" + "="*70)
    print("EDA COMPLETE")
    print("="*70)

# Run complete EDA
complete_eda(df, target_column='purchase_amount')
````
